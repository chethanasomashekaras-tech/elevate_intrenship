```python
import pandas as pd

# Load the dataset
df = pd.read_csv("/content/netflix.csv")

# Display the first 5 rows
df.head()
```

```python
print("Dataset loaded successfully!")
print("Rows and columns:", df.shape)
```

```python
df.info()
```

```python
# Check missing values in each column

missing_values = df.isnull().sum()

print("Missing values in each column:")
print(missing_values)
```

```python
# Check for duplicate rows

duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)
```

```python
# Display duplicate rows, if any

df[df.duplicated()]
```

```python
# Clean and standardize column names

df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

print("Cleaned column names:")
print(df.columns.tolist())
```

```python
df["release_year"]
```

```python
# Handle missing values

df["director"] = df["director"].fillna("Unknown")
df["cast"] = df["cast"].fillna("Unknown")
df["country"] = df["country"].fillna("Unknown")
df["rating"] = df["rating"].fillna("Unknown")
df["duration"] = df["duration"].fillna("Unknown")
```

```python
# Check missing values after cleaning

print(df.isnull().sum())
```

```python
# Convert date_added to datetime format

df["date_added"] = pd.to_datetime(
    df["date_added"],
    errors="coerce"
)

print("Data type of date_added:")
print(df["date_added"].dtype)
```

```python
# Display some dates

print(df["date_added"].head(10))
```

```python
print("Missing dates:", df["date_added"].isnull().sum())
```

```python
print("Missing dates:", df["date_added"].isnull().sum())
print("\nData type:", df["date_added"].dtype)
```

```python
# Load the original date column again for comparison

original_df = pd.read_csv("/content/netflix.csv")

print("Original missing dates:", original_df["date_added"].isnull().sum())

print("\nOriginal date examples:")
print(original_df["date_added"].head(20))
```

```python
# Find values that could not be converted to dates

invalid_dates = original_df[
    pd.to_datetime(original_df["date_added"], errors="coerce").isnull()
    & original_df["date_added"].notnull()
]

print("Non-empty dates that could not be converted:", len(invalid_dates))
print(invalid_dates["date_added"].unique())
```

```python
# Remove extra spaces from date_added

df["date_added"] = df["date_added"].astype("string").str.strip()

# Convert to datetime
df["date_added"] = pd.to_datetime(
    df["date_added"],
    errors="coerce"
)

# Check the result
print("Data type:", df["date_added"].dtype)
print("Missing dates:", df["date_added"].isnull().sum())
```

```python
# Reload the original dataset
df = pd.read_csv("/content/netflix.csv")

# Clean spaces from the date column
df["date_added"] = df["date_added"].str.strip()

# Convert dates to datetime
df["date_added"] = pd.to_datetime(
    df["date_added"],
    format="mixed",
    errors="coerce"
)

# Check the result
print("Data type:", df["date_added"].dtype)
print("Missing dates:", df["date_added"].isnull().sum())
```

```python
# Check the duration column

print(df["duration"].head(20))
print("\nData type:", df["duration"].dtype)
print("\nMissing values:", df["duration"].isnull().sum())
```

```python
# Check the rows where duration is missing

print(df[df["duration"].isnull()][["title", "type", "duration"]])
```

```python
# Replace missing duration values with "Unknown"

df["duration"] = df["duration"].fillna("Unknown")

print("Missing duration values:", df["duration"].isnull().sum())
```

```python
# Remove unnecessary spaces from text columns

text_columns = [
    "type",
    "title",
    "director",
    "cast",
    "country",
    "rating",
    "duration",
    "listed_in",
    "description"
]

for column in text_columns:
    df[column] = df[column].astype("string").str.strip()

print("Text columns cleaned successfully.")
```

```python
print(df["type"].value_counts())
```

```python
print(df["rating"].value_counts().head(20))
```

```python
# Check data types of all columns

print(df.dtypes)
```

```python
print("date_added type:", df["date_added"].dtype)
print("release_year type:", df["release_year"].dtype)
```

```python
# Final data quality check

print("===== FINAL DATA QUALITY CHECK =====")

# 1. Dataset size
print("\n1. Dataset shape:")
print(df.shape)

# 2. Missing values
print("\n2. Missing values:")
print(df.isnull().sum())

# 3. Duplicate rows
print("\n3. Duplicate rows:")
print(df.duplicated().sum())

# 4. Data types
print("\n4. Data types:")
print(df.dtypes)
```

```python
# Basic numerical summary

df.describe()
```

```python
# Save the cleaned dataset

output_path = "/content/netflix_titles_cleaned.csv"

df.to_csv(output_path, index=False)

print("Cleaned dataset saved successfully!")
print(output_path)
```
